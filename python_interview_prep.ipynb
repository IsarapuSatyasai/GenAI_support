# 1. Functions

A **function** is a reusable block of code that performs a specific task.

Functions help us avoid repetition, break large problems into smaller pieces, and make code easier to test and maintain.

### Basic syntax

```python
def function_name(parameters):
    # function body
    return result
```

**Parameter** = variable in the function definition.  
**Argument** = actual value passed when calling the function.  
**Return value** = value sent back using `return`.

````python
def greet(name):
    return f"Hello, {name}!"

message = greet("Anita")
print(message)
````

## Function arguments

Python supports:
- positional arguments,
- keyword arguments,
- default arguments,
- `*args` for variable positional arguments,
- `**kwargs` for variable keyword arguments.

````python
def introduce(name, age=18):
    return f"{name} is {age} years old."

print(introduce("Anita"))
print(introduce("Rahul", 25))
print(introduce(age=30, name="Priya"))

def add_numbers(*args):
    return sum(args)

print(add_numbers(10, 20, 30))

def show_profile(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

show_profile(name="Anita", role="Data Scientist", experience=3)
````

## Scope

Variables created inside a function normally have **local scope**. Variables outside have **global scope**.

For maintainable code, prefer passing values into functions and returning results rather than relying heavily on global state.

````python
course = "Python"

def show_course():
    lesson = "Functions"
    print(course)
    print(lesson)

show_course()
````

## Coding Problem — Student Grade Calculator

Create:
1. `calculate_average(marks)`
2. `get_grade(average)`
3. `display_result(name, marks)`

Grade rules:
- A: 90 or above
- B: 75–89
- C: 60–74
- D: 50–59
- F: below 50

Try solving it before viewing the solution.

````python
# Try your solution here.
````

### Solution

````python
def calculate_average(marks):
    return sum(marks) / len(marks)

def get_grade(average):
    if average >= 90:
        return "A"
    if average >= 75:
        return "B"
    if average >= 60:
        return "C"
    if average >= 50:
        return "D"
    return "F"

def display_result(name, marks):
    average = calculate_average(marks)
    grade = get_grade(average)
    print(f"Student: {name}")
    print(f"Average: {average:.2f}")
    print(f"Grade: {grade}")

display_result("Rahul", [85, 92, 78, 88, 90])
````

**Key takeaway:** Keep each function focused on one clear responsibility.

# 2. Dictionary

A **dictionary** stores data as key-value pairs.

```python
student = {"name": "Rahul", "age": 22, "marks": 85}
```

Common operations include `get()`, `keys()`, `values()`, `items()`, `pop()`, and `update()`.

Dictionaries are especially common with JSON, APIs, configuration, and metadata.

````python
student = {"name": "Rahul", "age": 22, "marks": 85}

print(student["name"])
print(student.get("department"))

student["marks"] = 90
student["department"] = "Computer Science"

for key, value in student.items():
    print(key, "->", value)
````

## Coding Problem — Word Frequency Counter

Given:

`python is easy and python is powerful`

Create a dictionary containing the frequency of every word.

Hint: `frequency.get(word, 0)`.

### Solution

````python
text = "python is easy and python is powerful"
words = text.lower().split()
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

print(frequency)
````

**Key takeaway:** Dictionaries are ideal when you need to associate keys with values or count/group information.

# 3. Object-Oriented Programming (OOP)

**Object-Oriented Programming** organizes software around objects containing data and behavior.

- **Class:** blueprint for objects.
- **Object:** instance of a class.
- **Attribute:** data stored on an object.
- **Method:** function inside a class.
- `__init__`: initializes an object.
- `self`: reference to the current object.

Important OOP ideas include encapsulation, inheritance, and polymorphism.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(f"Name: {self.name}")
        print(f"Marks: {self.marks}")

student = Student("Rahul", 85)
student.display()
````

## Inheritance and polymorphism

A child class can reuse and extend a parent class. Polymorphism allows different classes to expose the same method interface with different behavior.

````python
class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def speak(self):
        print("Dog says: Woof!")

class Cat(Animal):
    def speak(self):
        print("Cat says: Meow!")

for animal in [Dog(), Cat()]:
    animal.speak()
````

## Coding Problem — Bank Account

Create a `BankAccount` class with:
- `account_holder`
- `account_number`
- `balance`

Add:
- `deposit(amount)`
- `withdraw(amount)`
- `display_balance()`

Withdrawal should only happen when sufficient funds are available.

### Solution

````python
class BankAccount:
    def __init__(self, account_holder, account_number, balance=0):
        self.account_holder = account_holder
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("Insufficient balance")

    def display_balance(self):
        print(f"Account holder: {self.account_holder}")
        print(f"Balance: ₹{self.balance:.2f}")

account = BankAccount("Rahul", "ACC1001", 10000)
account.deposit(5000)
account.withdraw(3000)
account.display_balance()
````

**Key takeaway:** Use classes when your program has entities with state and behavior.

# 4. Exception Handling

An **exception** is a runtime event that interrupts normal program execution.

Common exceptions:
- `ValueError`
- `TypeError`
- `ZeroDivisionError`
- `FileNotFoundError`
- `KeyError`

Use `try`, `except`, `else`, and `finally` to handle errors safely. Use `raise` when your program needs to reject invalid data.

````python
try:
    number = int("100")
    result = 100 / number
except ValueError:
    print("Invalid number")
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("Result:", result)
finally:
    print("Execution finished")
````

## Raising exceptions

You can deliberately raise an exception when a business rule is violated.

````python
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    return age

print(set_age(25))
````

## Coding Problem — Safe Calculator

Create a calculator that:
- accepts two numbers,
- supports `+`, `-`, `*`, `/`,
- handles invalid numeric input,
- handles division by zero,
- handles invalid operators.

### Solution

````python
try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    operator = input("Enter operator (+, -, *, /): ")

    if operator == "+":
        result = num1 + num2
    elif operator == "-":
        result = num1 - num2
    elif operator == "*":
        result = num1 * num2
    elif operator == "/":
        result = num1 / num2
    else:
        raise ValueError("Invalid operator")

    print("Result:", result)

except ValueError as error:
    print("Error:", error)
except ZeroDivisionError:
    print("Cannot divide by zero")
````

**Key takeaway:** Catch specific exceptions when possible. Do not use exception handling to hide programming errors.

# 5. Files Handling

Python can read and write persistent data using files.

| Mode | Purpose |
|---|---|
| `r` | Read |
| `w` | Write/replace |
| `a` | Append |
| `x` | Create |

Prefer:

```python
with open("file.txt", "r") as file:
    content = file.read()
```

The `with` statement ensures the file is closed automatically.

````python
content = "INFO User logged in\nERROR Database connection failed\nINFO User logged out\nERROR File not found\nINFO User logged in\n"

with open("app.log", "w") as file:
    file.write(content)

with open("app.log", "r") as file:
    print(file.read())
````

## Reading and appending

Useful methods include `read()`, `readline()`, and `readlines()`. You can also iterate over a file line by line.

Appending with mode `a` adds content without replacing the file.

````python
with open("app.log", "a") as file:
    file.write("INFO User logged out\n")

with open("app.log", "r") as file:
    for line in file:
        print(line.strip())
````

## Coding Problem — Log Analyzer

Analyze `app.log` and count:
- Number of `INFO` messages.
- Number of `ERROR` messages.

### Solution

````python
info_count = 0
error_count = 0

with open("app.log", "r") as file:
    for line in file:
        if "INFO" in line:
            info_count += 1
        elif "ERROR" in line:
            error_count += 1

print("INFO:", info_count)
print("ERROR:", error_count)
````

## JSON files

JSON is widely used by APIs and AI applications.

```python
import json
json.dump(data, file)
data = json.load(file)
```

````python
import json

student = {
    "name": "Rahul",
    "age": 22,
    "skills": ["Python", "SQL", "Machine Learning"]
}

with open("student.json", "w") as file:
    json.dump(student, file, indent=4)

with open("student.json", "r") as file:
    loaded_student = json.load(file)

print(loaded_student)
````

**Key takeaway:** File handling is useful for logs, datasets, configuration, and persistent application data.

# 6. Iterators & Generators

An **iterable** is an object that can be iterated over, such as a list, tuple, string, dictionary, or set.

An **iterator** produces one item at a time. Use `iter()` to obtain one and `next()` to retrieve the next item.

When an iterator has no more values, Python raises `StopIteration`.

A **generator** is a convenient way to create an iterator. Generator functions use `yield` and produce values lazily.

````python
numbers = [10, 20, 30]
iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

## Generators

A generator pauses at `yield` and resumes from that point when asked for the next value.

This is useful when processing large or streaming data because values do not all need to exist in memory at the same time.

````python
def even_numbers(n):
    for number in range(1, n + 1):
        if number % 2 == 0:
            yield number

for number in even_numbers(10):
    print(number)
````

## Coding Problem — Square Number Generator

Create a generator that produces the first `n` square numbers.

For `n = 5`, expected output:

`1, 4, 9, 16, 25`

### Solution

````python
def square_numbers(n):
    for number in range(1, n + 1):
        yield number ** 2

for value in square_numbers(5):
    print(value)
````

**Key takeaway:** Generators are useful for large files, streams, batches, and memory-sensitive workloads.

# 7. Decorators

A **decorator** is a function that takes another function and adds behavior without changing the original function's source code.

Python functions are first-class objects: they can be stored in variables, passed as arguments, and returned from functions.

Common uses include logging, timing, authentication, caching, validation, and monitoring.

````python
def greet():
    return "Hello"

message_function = greet
print(message_function())
````

## Building a decorator

A decorator commonly contains a wrapper function:

```python
def decorator(function):
    def wrapper(*args, **kwargs):
        ...
    return wrapper
```

The `@decorator` syntax applies the decorator to a function.

````python
from functools import wraps

def log_call(function):
    @wraps(function)
    def wrapper(*args, **kwargs):
        print(f"Calling {function.__name__}")
        result = function(*args, **kwargs)
        print("Function completed")
        return result
    return wrapper

@log_call
def add(a, b):
    return a + b

print(add(10, 20))
````

## Coding Problem — Execution Timer

Create a decorator that:
1. records the start time,
2. runs the original function,
3. records the end time,
4. prints execution time,
5. returns the original result.

### Solution

````python
import time
from functools import wraps

def execution_timer(function):
    @wraps(function)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = function(*args, **kwargs)
        end = time.perf_counter()
        print(f"Execution time: {end - start:.6f} seconds")
        return result
    return wrapper

@execution_timer
def calculate_sum(n):
    return sum(range(n))

result = calculate_sum(1_000_000)
print("Result:", result)
````

**Key takeaway:** Decorators let you add reusable cross-cutting behavior such as logging and timing.

## Basic Pydantic model

A Pydantic model normally inherits from `BaseModel`.

```python
from pydantic import BaseModel

class Student(BaseModel):
    name: str
    age: int
```

````python
from pydantic import BaseModel

class StudentModel(BaseModel):
    name: str
    age: int
    marks: float

student = StudentModel(name="Rahul", age=22, marks=85.5)

print(student)
print(student.model_dump())
````

## Field validation

`Field()` lets us enforce constraints.

- `Field(gt=0, lt=100)` — greater than 0 and less than 100.
- `Field(ge=0, le=100)` — between 0 and 100 inclusive.

````python
from pydantic import BaseModel, Field

class StudentModel(BaseModel):
    name: str
    age: int = Field(gt=0, lt=100)
    marks: float = Field(ge=0, le=100)
    department: str

student = StudentModel(
    name="Rahul",
    age=22,
    marks=85.5,
    department="Computer Science"
)

print(student.model_dump())
````

## Validation errors

Invalid data raises `ValidationError`, which describes which fields failed validation.

````python
from pydantic import ValidationError

try:
    StudentModel(
        name="Rahul",
        age=-5,
        marks=120,
        department="Computer Science"
    )
except ValidationError as error:
    print(error)
````
